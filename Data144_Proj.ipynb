```python
import pandas as pd

df_train = pd.read_csv('data144_proj/train.csv')
df_test = pd.read_csv('data144_proj/test.csv')
```

```python
#TRAIN DATA CLEANING
df_train["Age"] = df_train["Age"].fillna(df_train["Age"].median())
df_train["Embarked"] = df_train["Embarked"].fillna(df_train["Embarked"].mode()[0])

df_train = df_train.drop(columns=["Cabin", "PassengerId", "Name", "Ticket"])

df_train["Sex"] = df_train["Sex"].map({"male": 0, "female": 1})
df_train = pd.get_dummies(df_train, columns=["Embarked"], dtype=int)
```

```python
# TEST DATA CLEANING

df_test["Age"] = df_test["Age"].fillna(df_train["Age"].median())
df_test["Fare"] = df_test["Fare"].fillna(df_train["Fare"].median())

df_test = df_test.drop(columns=["Cabin", "PassengerId", "Name", "Ticket"])

df_test["Sex"] = df_test["Sex"].map({"male": 0, "female": 1})
df_test = pd.get_dummies(df_test, columns=["Embarked"], dtype=int)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# MODEL TRAINING
X = df_train.drop(columns=["Survived"])
y = df_train["Survived"]

model = RandomForestClassifier(random_state=42)

param_grid = {
    "n_estimators": [100, 300,1200],
    "max_depth": [None, 10, 20],
    "min_samples_leaf": [1, 2, 3, 4, 5],
    "max_features": ["sqrt", "log2", None],
    "criterion": ["gini", "entropy", "log_loss"]}

grid = GridSearchCV(
    model,
    param_grid,
    cv=5,
    scoring="accuracy"
)


grid.fit(X, y)
```

```python
# PREDICT TEST DATA
pred = grid.predict(df_test)
pred
```

```python
test_passenger = pd.read_csv('data144_proj/test.csv')
```

```python
submission = test_passenger[["PassengerId"]]
submission["Survived"] = pred
submission
```

```python
submission.to_csv("data144_proj/submission.csv", index=False)
```
